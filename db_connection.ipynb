```python
vibrationData = vibrationData.sort_values(by="_time").reset_index(drop=True)
waMaData = waMaData.sort_values(by="_time").reset_index(drop=True)
```

```python
data = pd.merge_asof(vibrationData, waMaData, on="_time", direction="backward", tolerance=pd.Timedelta("2s"))
```

```python
display(data)
```

```python
data.columns
```

```python
cols_to_drop = ['result_x', 'table_x', '_start_x', '_stop_x', '_measurement_x', 'host_x', 'path', 'sensor_id',
                'result_y', 'table_y', '_start_y', '_stop_y', '_measurement_y','host_y', 'src']

data.drop(columns=cols_to_drop, inplace=True)
```

```python
data.info()
```

```python
waMaColumns = ['active', 'analogIn0', 'flagWord4', 'inputWord0', 'outputWord0', 'remainingTime', 'stepNr']
```

```python
data[waMaColumns] = data[waMaColumns].ffill()
```

```python
data
```

```python
valid_wama_cols = [col for col in waMaColumns if col in data.columns]
if valid_wama_cols:
    # Identify rows with NaNs in the specified WaMa columns
    drop_mask = data[valid_wama_cols].isna().any(axis=1)

    if drop_mask.any():
        # Create a temporary copy to group contiguous blocks of dropped rows
        temp_df = data.copy()
        temp_df['_drop_block'] = (drop_mask != drop_mask.shift()).cumsum()

        print(f"Debug: Total rows to delete: {drop_mask.sum()}")
        print("Debug: Identified the following time ranges for deletion:")

        for _, group in temp_df[drop_mask].groupby('_drop_block'):
            start_time = group['_time'].min()
            end_time = group['_time'].max()
            print(f" - {start_time} to {end_time} ({len(group)} rows)")
        data = temp_df[~drop_mask].drop(columns=['_drop_block'])
    else:
        print("Debug: No rows found with missing WaMa values.")
```

```python
print(f"Original waMa data shape: {waMaData.shape}")
print(f"Original vibration data shape: {vibrationData.shape}")
print(f"Merged dataset shape: {data.shape}")
print(f"NaN count in waMa columns: {data[waMaColumns].isna().sum().sum()}")
```

```python
colsToCompare = ["inputWord0", "outputWord0", "flagWord4"]
```

```python
for c in colsToCompare:
    prev_val = data[c].shift(1).fillna(0).astype(int)
    data[f"{c}_changed_bits"] = data[c].astype(int) ^ prev_val
```

```python
data
```

```python
for c in colsToCompare:
    prev_val = data[c].shift(1).fillna(0).astype(int)
    for i in range(16):  # Assuming 16-bit integers
        data[f'{c}_bit_{i}_changed'] = (data[f'{c}_changed_bits'] & (1 << i)) != 0
        data[f'{c}_bit_{i}'] = (data[c].astype(int) & (1 << i)) != 0
```
